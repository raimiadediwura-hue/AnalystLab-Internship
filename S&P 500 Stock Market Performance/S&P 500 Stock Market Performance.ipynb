```python
# extract year
stock_df['year'] = stock_df['date'].dt.year

stock_df[['date', 'year']].head()
```

```python
# extract month 
stock_df['month'] = stock_df['date'].dt.month

stock_df[['date', 'month']].head()
```

```python
# Daily price change
stock_df['daily_price_change'] = (
    stock_df.groupby('symbol')['close'].diff()
)
```

```python
# Moving average 
stock_df['MA7'] = (
    stock_df.groupby('symbol')['close']
    .transform(lambda x: x.rolling(window=7, min_periods=7).mean())
)
```

```python
stock_df['MA20'] = (
    stock_df.groupby('symbol')['close']
    .transform(lambda x: x.rolling(window=20, min_periods=20).mean())
)
```

```python
stock_df['MA50'] = (
    stock_df.groupby('symbol')['close']
    .transform(lambda x: x.rolling(window=50, min_periods=50).mean())
)
```

```python
# question 1
# Highest closing price for each stock

highest_closing_prices = (
    stock_df.groupby('symbol')['close']
    .max()
    .sort_values(ascending=False)
)

# Display top 10 stocks
highest_closing_prices.head(10)
```

```python
# question 2
# Calculate first and last closing prices for each stock
stock_returns = (
    stock_df.sort_values(['symbol', 'date'])
    .groupby('symbol')
    .agg(
        First_Close=('close', 'first'),
        Last_Close=('close', 'last'),
        Start_Date=('date', 'min'),
        End_Date=('date', 'max')
    )
)

# Calculate cumulative percentage return
stock_returns['Cumulative_Return_%'] = (
    (stock_returns['Last_Close'] - stock_returns['First_Close'])
    / stock_returns['First_Close']
) * 100

# Rank stocks from highest to lowest return
highest_returns = stock_returns.sort_values(
    'Cumulative_Return_%',
    ascending=False
)

# Rank stocks from lowest to highest return
lowest_returns = stock_returns.sort_values(
    'Cumulative_Return_%',
    ascending=True
)
```

```python
top_returns = highest_returns.head(10)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_returns['Cumulative_Return_%'],
    y=top_returns.index,
    hue=top_returns.index,
    palette='Greens',
    legend=False
)

plt.title('Top 10 Stocks with the Highest Cumulative Returns (2014–2017)')
plt.xlabel('Cumulative Return (%)')
plt.ylabel('Stock Symbol')
plt.show()
```

```python
bottom_returns = lowest_returns.head(10)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=bottom_returns['Cumulative_Return_%'],
    y=bottom_returns.index,
    hue=bottom_returns.index,
    palette='Reds',
    legend=False
)

plt.title('Top 10 Stocks with the Lowest Cumulative Returns (2014–2017)')
plt.xlabel('Cumulative Return (%)')
plt.ylabel('Stock Symbol')
plt.show()
```

```python
dashboard_df = stock_df.copy()

dashboard_df['date'] = pd.to_datetime(dashboard_df['date'])

dashboard_df = dashboard_df.sort_values(
    ['symbol', 'date']
).reset_index(drop=True)

dashboard_df.head()
```

```python
selected_stocks = ['AAPL', 'MSFT', 'AMZN', 'GOOGL', 'JPM']
```

```python
# calculate kpi's

total_symbols = dashboard_df['symbol'].nunique()

total_records = len(dashboard_df)

start_date = dashboard_df['date'].min().strftime('%d %b %Y')

end_date = dashboard_df['date'].max().strftime('%d %b %Y')

average_closing_price = (
    dashboard_df[
        dashboard_df['symbol'].isin(selected_stocks)
    ]['close'].mean()
)

print("Total Stock Symbols:", total_symbols)
print("Total Trading Records:", f"{total_records:,}")
print("Available Date Range:", start_date, "to", end_date)
print("Average Closing Price:", f"${average_closing_price:,.2f}")
```

```python
kpi_fig = go.Figure()

kpi_values = [
    ("Total Stock Symbols", f"{total_symbols:,}"),
    ("Total Trading Records", f"{total_records:,}"),
    ("Available Date Range", f"{start_date}<br>to {end_date}"),
    ("Average Closing Price", f"${average_closing_price:,.2f}")
]

for i, (title, value) in enumerate(kpi_values):
    kpi_fig.add_trace(
        go.Indicator(
            mode="number",
            value=(
                total_symbols if i == 0 else
                total_records if i == 1 else
                average_closing_price if i == 3 else 0
            ),
            title={"text": title},
            number={
                "prefix": "$" if i == 3 else "",
                "valueformat": ",.2f" if i == 3 else ",.0f"
            },
            domain={
                "row": 0,
                "column": i
            }
        )
    )

kpi_fig.update_layout(
    grid={"rows": 1, "columns": 4, "pattern": "independent"},
    height=220,
    title="S&P 500 Market Overview - Key Indicators"
)

kpi_fig.show()
```

```python
def create_returns_dashboard(selected_symbols, start_date, end_date):

    if not selected_symbols:
        print("Please select at least one stock.")
        return

    if start_date is None or end_date is None:
        print("Please select both dates.")
        return

    if start_date > end_date:
        print("Start date cannot be later than end date.")
        return

    # Filter selected stocks and dates
    filtered_df = returns_df[
        (returns_df['symbol'].isin(selected_symbols)) &
        (returns_df['date'] >= pd.Timestamp(start_date)) &
        (returns_df['date'] <= pd.Timestamp(end_date))
    ].copy()

    if filtered_df.empty:
        print("No data available for the selected filters.")
        return

    # Calculate comparison metrics for the selected period
    summary = (
        filtered_df.groupby('symbol')
        .agg(
            Average_Daily_Return=('daily_return', 'mean'),
            Daily_Volatility=('daily_return', 'std'),
            Highest_Daily_Return=('daily_return', 'max'),
            Lowest_Daily_Return=('daily_return', 'min'),
            Trading_Records=('daily_return', 'count')
        )
        .reset_index()
    )

    summary['Annualised_Volatility'] = (
        summary['Daily_Volatility'] * np.sqrt(252)
    )

    # Create dashboard layout
    fig = make_subplots(
        rows=3,
        cols=2,
        specs=[
            [{"type": "xy", "colspan": 2}, None],
            [{"type": "xy"}, {"type": "xy"}],
            [{"type": "xy"}, {"type": "xy"}]
        ],
        subplot_titles=[
            "Daily Percentage Returns",
            "Average Daily Returns by Stock",
            "Distribution of Daily Returns",
            "Historical Volatility",
            "Annualised Volatility Comparison"
        ],
        vertical_spacing=0.14,
        horizontal_spacing=0.12
    )

    # 1. Daily percentage returns
    for symbol in selected_symbols:

        stock_data = filtered_df[
            filtered_df['symbol'] == symbol
        ]

        fig.add_trace(
            go.Scatter(
                x=stock_data['date'],
                y=stock_data['daily_return'],
                mode='lines',
                name=symbol
            ),
            row=1, col=1
        )

    # 2. Average daily returns
    fig.add_trace(
        go.Bar(
            x=summary['symbol'],
            y=summary['Average_Daily_Return'],
            name='Average Daily Return'
        ),
        row=2, col=1
    )

    # 3. Return distribution
    for symbol in selected_symbols:

        stock_data = filtered_df[
            filtered_df['symbol'] == symbol
        ]

        fig.add_trace(
            go.Box(
                y=stock_data['daily_return'],
                name=symbol,
                boxmean=True
            ),
            row=2, col=2
        )

    # 4. Historical daily volatility
    fig.add_trace(
        go.Bar(
            x=summary['symbol'],
            y=summary['Daily_Volatility'],
            name='Daily Volatility'
        ),
        row=3, col=1
    )

    # 5. Annualised volatility comparison
    fig.add_trace(
        go.Bar(
            x=summary['symbol'],
            y=summary['Annualised_Volatility'],
            name='Annualised Volatility'
        ),
        row=3, col=2
    )

    # Reference line for zero daily return
    fig.add_hline(
        y=0,
        line_dash='dash',
        line_color='black',
        row=1, col=1
    )

    fig.add_hline(
        y=0,
        line_dash='dash',
        line_color='black',
        row=2, col=1
    )

    fig.update_layout(
        title={
            'text': (
                "S&P 500 Returns and Volatility  & "
                "Predictive Insights"
                f"<br><sup>Page 3: Returns and Volatility | "
                f"{start_date} to {end_date}</sup>"
            ),
            'x': 0.5
        },
        template='plotly_white',
        height=1200,
        hovermode='x unified',
        showlegend=True,
        legend=dict(
            orientation='h',
            yanchor='bottom',
            y=1.02,
            xanchor='right',
            x=1
        )
    )

    fig.update_xaxes(title_text='Date', row=1, col=1)
    fig.update_yaxes(title_text='Daily Return (%)', row=1, col=1)

    fig.update_xaxes(title_text='Stock Symbol', row=2, col=1)
    fig.update_yaxes(title_text='Average Return (%)', row=2, col=1)

    fig.update_yaxes(title_text='Daily Return (%)', row=2, col=2)

    fig.update_xaxes(title_text='Stock Symbol', row=3, col=1)
    fig.update_yaxes(title_text='Daily Volatility (%)', row=3, col=1)

    fig.update_xaxes(title_text='Stock Symbol', row=3, col=2)
    fig.update_yaxes(title_text='Annualised Volatility (%)', row=3, col=2)

    fig.show()

    display(
        summary.round(4).sort_values(
            'Average_Daily_Return',
            ascending=False
        )
    )
```

```python
def create_volume_dashboard(selected_symbols, start_date, end_date):

    # Check stock selection
    if not selected_symbols:
        print("Please select at least one stock symbol.")
        return

    # Check date selection
    if start_date is None or end_date is None:
        print("Please select both start and end dates.")
        return

    if start_date > end_date:
        print("Start date cannot be later than end date.")
        return

    # Filter selected stocks
    filtered_df = volume_df[
        volume_df['symbol'].isin(selected_symbols)
    ].copy()

    # Apply date filters
    filtered_df = filtered_df[
        (filtered_df['date'] >= pd.Timestamp(start_date)) &
        (filtered_df['date'] <= pd.Timestamp(end_date))
    ]

    if filtered_df.empty:
        print("No records found for the selected filters.")
        return

    # -----------------------------------
    # 1. Calculate average trading volume
    # -----------------------------------

    average_volume = (
        filtered_df.groupby('symbol')['volume']
        .mean()
        .reset_index(name='Average_Daily_Volume')
    )

    # -----------------------------------
    # 2. Identify highest and lowest volume observations
    # -----------------------------------

    highest_volume = (
        filtered_df.nlargest(5, 'volume')[
            ['symbol', 'date', 'volume', 'close', 'daily_return']
        ]
    )

    lowest_volume = (
        filtered_df.nsmallest(5, 'volume')[
            ['symbol', 'date', 'volume', 'close', 'daily_return']
        ]
    )

    # -----------------------------------
    # 3. Create subplot dashboard
    # -----------------------------------

    fig = make_subplots(
        rows=2,
        cols=2,
        specs=[
            [{"type": "xy"}, {"type": "xy"}],
            [{"type": "table"}, {"type": "xy"}]
        ],
        subplot_titles=(
            "Trading Volume Trends",
            "Average Daily Trading Volume",
            "Highest and Lowest Volume Observations",
            "Trading Volume vs Price Movement"
        ),
        vertical_spacing=0.15,
        horizontal_spacing=0.12
    )

    # -----------------------------------
    # Chart 1: Trading volume trends
    # -----------------------------------

    for symbol in selected_symbols:

        stock_data = filtered_df[
            filtered_df['symbol'] == symbol
        ]

        fig.add_trace(
            go.Scatter(
                x=stock_data['date'],
                y=stock_data['volume'],
                mode='lines',
                name=symbol,
                hovertemplate=(
                    'Date: %{x|%Y-%m-%d}<br>'
                    'Volume: %{y:,.0f}<extra>%{fullData.name}</extra>'
                )
            ),
            row=1,
            col=1
        )

    # -----------------------------------
    # Chart 2: Average trading volume
    # -----------------------------------

    fig.add_trace(
        go.Bar(
            x=average_volume['symbol'],
            y=average_volume['Average_Daily_Volume'],
            name='Average Volume',
            marker=dict(
                line=dict(width=0.5)
            ),
            hovertemplate=(
                'Stock: %{x}<br>'
                'Average Volume: %{y:,.0f}<extra></extra>'
            )
        ),
        row=1,
        col=2
    )

    # -----------------------------------
    # Chart 3: Highest and lowest observations
    # -----------------------------------

    extreme_records = pd.concat([
        highest_volume.assign(Volume_Category='Highest'),
        lowest_volume.assign(Volume_Category='Lowest')
    ])

    table_values = [
        extreme_records['symbol'],
        extreme_records['date'].dt.strftime('%Y-%m-%d'),
        extreme_records['Volume_Category'],
        extreme_records['volume'].map(lambda x: f'{x:,.0f}'),
        extreme_records['daily_return'].map(
            lambda x: f'{x:.2f}%' if pd.notna(x) else 'N/A'
        )
    ]

    fig.add_trace(
        go.Table(
            header=dict(
                values=[
                    'Stock',
                    'Date',
                    'Category',
                    'Volume',
                    'Daily Return'
                ],
                align='center'
            ),
            cells=dict(
                values=table_values,
                align='center'
            )
        ),
        row=2,
        col=1
    )

    # -----------------------------------
    # Chart 4: Trading volume vs price movement
    # -----------------------------------

    scatter_df = filtered_df.dropna(
        subset=['volume', 'absolute_daily_return', 'daily_return']
    )

    for symbol in selected_symbols:

        stock_data = scatter_df[
            scatter_df['symbol'] == symbol
        ]

        fig.add_trace(
            go.Scatter(
                x=stock_data['volume'],
                y=stock_data['absolute_daily_return'],
                mode='markers',
                name=symbol,
                marker=dict(
                    size=7,
                    opacity=0.6,
                    color=stock_data['daily_return'],
                    colorscale='RdYlGn',
                    showscale=False
                ),
                customdata=stock_data[['daily_return', 'date']].to_numpy(),
                hovertemplate=(
                    'Stock: ' + symbol +
                    '<br>Volume: %{x:,.0f}' +
                    '<br>Absolute Daily Return: %{y:.2f}%' +
                    '<br>Daily Return: %{customdata[0]:.2f}%' +
                    '<br>Date: %{customdata[1]|%Y-%m-%d}'
                    '<extra></extra>'
                )
            ),
            row=2,
            col=2
        )

    # -----------------------------------
    # Dashboard layout
    # -----------------------------------

    fig.update_layout(
        title=dict(
            text='Trading Volume Analysis',
            x=0.5,
            xanchor='center'
        ),
        height=950,
        template='plotly_white',
        showlegend=True,
        legend=dict(
            orientation='h',
            y=-0.12
        )
    )

    fig.update_xaxes(
        title_text='Date',
        row=1,
        col=1
    )

    fig.update_yaxes(
        title_text='Trading Volume (Shares)',
        row=1,
        col=1
    )

    fig.update_xaxes(
        title_text='Stock Symbol',
        row=1,
        col=2
    )

    fig.update_yaxes(
        title_text='Average Daily Volume (Shares)',
        row=1,
        col=2
    )

    fig.update_xaxes(
        title_text='Trading Volume (Shares)',
        type='log',
        row=2,
        col=2
    )

    fig.update_yaxes(
        title_text='Absolute Daily Return (%)',
        row=2,
        col=2
    )

    # Display dashboard
    fig.show()

    # -----------------------------------
    # Display additional summary
    # -----------------------------------

    print("\nTRADING VOLUME SUMMARY")

    summary = (
        filtered_df.groupby('symbol')
        .agg(
            Total_Trading_Volume=('volume', 'sum'),
            Average_Daily_Volume=('volume', 'mean'),
            Highest_Daily_Volume=('volume', 'max'),
            Lowest_Daily_Volume=('volume', 'min'),
            Trading_Records=('volume', 'count')
        )
        .reset_index()
    )

    display(summary.round(2))
```

```python
dashboard_output = widgets.interactive_output(
    create_volume_dashboard,
    {
        'selected_symbols': symbol_selector,
        'start_date': start_date_picker,
        'end_date': end_date_picker
    }
)

display(dashboard_output)
```

```python
def save_volume_dashboard(selected_symbols, start_date, end_date):

    if not selected_symbols or start_date is None or end_date is None:
        print("Select stocks and dates before exporting.")
        return

    filtered_df = volume_df[
        (volume_df['symbol'].isin(selected_symbols)) &
        (volume_df['date'] >= pd.Timestamp(start_date)) &
        (volume_df['date'] <= pd.Timestamp(end_date))
    ].copy()

    if filtered_df.empty:
        print("No records available for export.")
        return

    fig = px.line(
        filtered_df,
        x='date',
        y='volume',
        color='symbol',
        title='Trading Volume Analysis',
        labels={
            'date': 'Date',
            'volume': 'Trading Volume (Shares)',
            'symbol': 'Stock Symbol'
        }
    )

    fig.update_layout(
        template='plotly_white',
        title_x=0.5
    )

    fig.write_html(
        'Trading_Volume_Analysis.html',
        include_plotlyjs=True
    )

    print("Dashboard saved successfully.")
```

```python
save_volume_dashboard(
    tuple(symbol_selector.value),
    start_date_picker.value,
    end_date_picker.value
)
```
